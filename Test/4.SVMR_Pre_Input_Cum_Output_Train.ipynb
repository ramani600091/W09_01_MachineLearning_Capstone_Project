# Training Model

```python
import pickle
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
```

```python

dataset = pd.read_csv("HeartFailureDataset v2.csv",index_col=None)
dataset = pd.get_dummies(dataset,dtype=int, drop_first=True)
dataset = dataset.fillna(0)

#Converting age
dataset['age'] = (dataset['age'] / 365).astype(int)
print(dataset)
independent=dataset[['age','height','weight','ap_hi','ap_lo','cholesterol']]
dependent  =dataset['active']

## Check for negative values
neg_rows = independent[(independent < 0).any(axis=1)]
print("--Negative Columns:\n",neg_rows)
print("MINIMUM VALUES: \n",independent.min())


```

# Splitting Data into Training and Testing Sets

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(independent, dependent, test_size=0.30, random_state=0)
```

# Data Preprocessing and Model Training

```python
###Pre-processing the INPUT
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
# y_pred = regressor.predict(X_test)
# print(X_train,y_train)

###Pre-processing the OUTPUT

# scy = StandardScaler()
# y_train = scy.fit_transform(y_train)
# y_test = scy.transform(y_test)


```

```python

from sklearn.svm import SVC
regressor = SVC(kernel="linear", random_state=0)
regressor.fit(X_train, y_train)

# regressor.support_
y_pred = regressor.predict(X_test)
```

# Model Saving, Loading, and Prediction

```python
import pickle
filename     = "finalized_model_svr.sav"
pickle.dump(regressor, open(filename, 'wb'))
preinput     = sc.transform([[47, 156, 56, 100, 60,1]])
loaded_model = pickle.load(open("finalized_model_svr.sav", "rb"))
result       = loaded_model.predict(preinput)
# preoutput    = scy.inverse_transform([result])
```
